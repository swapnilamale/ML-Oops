```
# What is git Repo/Repository?
  Git Repo : A workspace folder which tracks changes in files within folder
             Every Git Repo has its own history which save's checkpoints of code called git commits
             folder labeled as git repo when it contains ".git" file which is created by "git init" command
             if we delete .git folder, repo become normal local directory
             it is not recommeded to have git repo under another git repo
```

```
There are three main areas in Git Repo:
 1) Working Directory → This is where you create and modify files.
    Files here can be:
        Untracked (new files Git is not yet aware of)
        Modified (changes made to tracked files but not yet staged)
        Deleted (Files were previously tracked but now its deleted)
        
 2) Staging Area (Index) → This is an intermediate area where you add changes using git add.
    Here, Git becomes aware of the changes and marks them as ready to be committed.
    
 3) Repository (.git folder) → This is where Git permanently stores committed snapshots of your project history.
 
 
                                Consider this as :
                                Working Directory → “I changed something”
                                Staging Area      → “Git, please be aware about it”
                                Repository        → “Saved permanently”
```

</div>

[gitignore.io](https://www.toptal.com/developers/gitignore/)

```
- .gitignore tells Git which files and folders should NOT be pushed to remote and should NOT be tracked 
- Git does not track EMPTY FOLDERS by default.
- If a folder contains only ignored files, Git treats it as empty.
- Because of this, such folders will not appear in the remote repository (GitHub).
- To keep the folder structure in the repository, add a placeholder file inside the folder.
- Common practice is to use a file named .gitkeep for this purpose.

Note : - .gitignore is usually created in the root directory of a Git repository, 
         which is the folder where the .git/ directory exists. 
       - A .gitignore file placed at the root applies to the entire repository, including all subfolders. 
       - If needed, additional .gitignore files can also be placed inside specific subfolders 
         to define folder-specific ignore rules.
```

#### .gitignore file syntax

```
                                # ignoring single file (<fila_name>)
                                    .DS_store
                                    /data/input_data/locality_mapping.pkl
                                
                                # ignoring entire directory (<folder_name>/)  
                                    temp/        
                                    build/  
                                    __pycache__/
                                    .venv/  
                                    ./
                                    ../
                                    *.egg-info/
                                
                                # ignoring entire .extention files (*.<extention>)   
                                    *.log       
                                    *.ipynb     
                                    *.ipynb_checkpoints
                                    *.pyc
                                    *.swp
                                    *.tar
                                    *.py[cod]
                                   
```

#### .gitkeep

``` - .gitkeep is just a normal hidden file.
    - It has no special meaning or power in Git.
    - Git does not treat .gitkeep differently from any other file.
    - It is only used by convention to keep an otherwise empty folder in the repository.
    - If .gitkeep matches a rule in .gitignore, it will be ignored like any other file.
    - Whether it is tracked or ignored depends entirely on .gitignore patterns.

```

#### start tracking ignore file mentioned in .gitignore

| Command                          | Description                                                                 |
|----------------------------------|-----------------------------------------------------------------------------|
| `git add -f <ignored_file>`     | Forces Git to add a file that matches a rule in `.gitignore` and starts tracking it from that point onward. |


#### Ignoring  a Previously Tracked File & Removing from Git Remote While Retaining It Locally

```
Scenario
A file or folder was previously tracked in a Git repository. It has now been added to .gitignore.
Adding a file to .gitignore does not stop tracking it if it was already tracked.
.gitignore only prevents new untracked files from being added.
If the file is already part of the repository history, Git will continue tracking it.
Now,
How to Keep the File Locally While Removing It from Remote and Ensuring It Is Not Pushed in the Next Commit?
To keep the file locally while removing it from version control and ensuring it is deleted from the remote repository in the next commit, the file must be removed from Git’s index (staging area) but not from the working directory.
```

| Command                     | Description                                                                 |
|:-----------------------------:|:-----------------------------------------------------------------------------:|
| `git rm --cached <file>`   | Removes the file from Git’s index (staging area) but keeps it in the working directory. </div><div>Stops tracking the file in future commits. |
                            

```                           
Why This Works
    --cached removes the file only from the index.
    The file remains physically present in the local working directory.
    Git stages the file as deleted.
    After commit and push, the file is removed from the remote repository.
    Since the file is now listed in .gitignore, future changes will not be tracked.
    
Note: This process does not remove the file from previous commit history. It only stops tracking it going forward.
```

    <h3>git branch and head pointer</h3>
</div>

```
- Think of branches as alternative timelines for a project
- They enable us to create separate contexts where we can try new things, try something experimental
  or even work on multiple ideas in parallel, without distubing the original working workflow
- if we make changes on one branch, they do not impact the other branches (Unless we merge the changes)
- we can merge thoes changes to main working flow once tried and tested
```

```
master/main : - In git we are always working on a branch, default branch name is master
                Note: this default master branch is not created when you initialise git repo
                      it is created when you do your first commit
              - this branch is created by git for user, any other braches needs to create by user only
              - this branch is also cosidered as official "sourch of truth" code, which is in or will go in production
              - in 2020 github renamed the default branch from "master" to "main", 
                however for git default branch name is still "master"
              - At any given moment only one git branch can be acitve given git repository 
                which is master/main initially and only the branch created by .git for user    
```

```
HEAD Pointer: 
- HEAD is a pointer that points to the currently active branch.
- The HEAD pointer always refers to one commit hash at a time, which you can see using git log or git reflog.
- The working directory reflects the state of the commit where HEAD is pointing.
- HEAD indicates which branch is currently active for committing. 
  By default, HEAD points to the tip (latest commit) of the active branch.
  When you switch branches, HEAD moves to the tip of the newly checked-out branch.
- When HEAD is not pointing to the tip of any branch and instead points directly to a commit, 
  it is called a detached HEAD state.
- All this information is stored inside the .git directory 

```

```
### HEAD vs Git Branch ####

HEAD and Git branches are two different pointers in Git.
Git Branch
- A Git branch is a movable pointer to a commit, When a new commit is created, 
  the current branch pointer moves forward. 
  i.e git branch pointer is defined only for that specific branch and 
      always points to tip (latest commit) of that branch 
- this is not movable pointer
- Multiple git branch can exist in a repository as multiple branches can exist 

HEAD
- HEAD is a special pointer that represents the current checkout position which relects current acitve working directory
- It usually points to a git branch pointer (which is by default tip of the branch), not directly to a commit.
- Only one HEAD exists at any time and can be moved to other git branch/commits using git checkout/switch command
- If HEAD points directly to a commit instead of a git branch, it is called a detached HEAD state.
```

```
- When you try to switch branches while having uncommitted changes,
  Git may block the checkout if those changes could conflict with files in the target branch.
- If there’s no direct conflict, Git will carry those changes to the new branch as unstaged changes, 
  which is often not what you want.
- To avoid both situations, git stash is used to temporarily save your working changes 
  without creating unnecessary commits.
- git stash is a super useful command that lets you park your incomplete work safely, 
  switch branches freely, and come back to it later whenever needed.
```

    <h3>Git Rebasing as Alternative to Merging</h3git restore
</div>

```
- git rebase moves your commits and reapplies them on top of another base commit.
- It rewrites commit history (unlike merge).
- rebase moves the entire feature branch so that it BEGINS at the tip of the master/main branch
- Think it like this : Replay my work as if I started from the latest code of master/main branch
- Why use rebase?
    To keep linear & clean history
    To avoid extra merge commits
    To update your branch with latest main before merging
- Rebasing is NOT recommended if the feature branch is already pushed to github and shared with other collaborators.
```

```
                                          STEP 1 : git checkout <feature_branch>
                                          STEP 2 : git rebase main

                    Takes commits from feature branch, Replays them on top of main brach
                    
                    
            A---B---C---D          (main)
                     \                                ------------>         A---B---C---D---E'---F'---G'   (feature)
                      E---F---G   (feature)    
                                                                              

Note : After rebasing, the feature branch commit hashes change, while the main branch commit hashes remain the same,
       because new commits are created on top of the main branch that contain the same changes 
       as the original feature branch commits.                                                                     
```

#### Rebasing with conflicts

```
During a rebase, you can either abort to revert changes or resolve conflicts and continue applying commits one by one.
1) Undo Rebasing Process
    git rebase --abort      

2) Resolve all the conflicts and run
    git add <file>
    git rebase --continue    
```

            - git is a KEY-VALUE data store or database where key came from hash function (SHA-1)
              and values are stored in objects(commit,tree,blob)
            - we can insert any kind of content into a Git repo and 
              Git will give us back a unique key we can later use to retrieve that content
  
  
Blob (Binary Large Object)
    - Blobs are the basic building block objects in Git.
    - A blob object is used to store the CONTENTS OF FILE(of any kind).
    - Blobs do not store filenames or directory structure, They contain only the RAW FILE CONTENT.
    - Each blob is identified by a unique SHA-1 hash, which acts as its key.
    
Tree (Git Tree Object)
    - A tree is a Git object used to represent the CONTENTS OF DIRECTORY.
    - A tree contains pointers (references) to:
          Blob objects (files)
          Other tree objects (subdirectories)
    - Each entry inside a tree stores:
          The SHA-1 hash of the referenced blob or tree
          The file mode (permissions/type)
          The object type (blob or tree)
          The filename
    A tree object organizes blobs and sub-trees to form the repository’s directory structure.
  
Commit (Git Commit Object)
    - A commit object combines a tree object with additional metadata.
    - It represents a snapshot of the project at a specific point in time.
    - A commit stores:
          A reference (SHA-1 hash) to the root tree object : a snapshot of code
          Reference(s) to parent commit(s) (none for first commit, multiple for merge commits)
          The author information
          The committer information
          The commit message
          Timestamp details
    - Commits create the history (linked list/graph) of the repository by pointing to their parent commits.
    
Annotated Tag (Git Tag Object)
    - An annotated tag is a pointer to a specific commit.
    - Unlike a lightweight tag, it is stored as a separate Git object.
    - It contains additional metadata about the tag.
          The commit hash it points to
          The tag name
          The tagger (author of the tag)
          The date/time of tagging
          An optional tag message
          Optional GPG signature

    - Structure relationship:
        Tag → Commit → Tree → Blob
```

***

***

***

***

***

***

***

***

***

***

***

***

***

***

***

<div style="text-align: center; color: blue;">
    <h1>Chapter 1: Github Basics</h1>
</div>

```

# Can One Local Git Repository Be Connected to Multiple GitHub Repositories?
    Many developers think one local Git repository can be connected to only one GitHub repository.That’s not true.
    A single local Git repository can be connected to multiple remote repositories at the same time.

# What does this mean?
    Your local project folder can:
        - Push code to GitHub Repository A
        - Push the same code to GitHub Repository B
        - Pull updates from different repositories
    In Git, these connections are called remotes.
    A) You can check your remotes using:
         git remote -v
    B) You can even add multiple remotes like this:
         git remote add origin <repo1_url>
         git remote add backup <repo2_url>
    C) After that, you can push to each one separately:
         git push origin main
         git push backup main

# Important Point
    Git does not automatically sync all remotes.
    If you push to one remote, it will NOT automatically push to the others.
    You must push to each remote intentionally.

When you clone a repository from remote that has multiple branches, Git does four important things:
    1. It downloads all branches and all commits from the remote repository.
    2. git sets default remote name as origin
    3. It creates something called extra remote-tracking branches for all remote branches
       having name <remote_name>/<branch_name> (ex: origin/main, origin/dev, etc.).
    4. It pops out only the default branch (main/master) as your active local branch.
    
- so once you are connected with remote and interacted by clone/fetch/pull/push command
  you will see two know of git branches
        1. Local Branches
            git branch : main
        2. Remote-Tracking Branches
            git branch -r : origin/main
                            origin/dev
                            origin/feature_x

 - Remote-tracking branches has only siginficance on local repository, they are not exist on remote
   These are not editable directly.
   They are snapshots of what those branches looked like on the remote the last time you interacted with remote.
 - remote-tracking branches are NOT a real local branch, they are just a pointer, 
   We can convert remote-tracking-branches to local to trace and make it editible
         git checkout -b <local_branch_name> <remote>/<remote_branch>
         git checkout -b dev origin/dev
 - Remote-tracking branches are created per remote, and each remote maintains its own set of remote branches

Note : - Git remote-tracking-branches are also automatically created/updated, when you push code to remote 
         with repect the that branch and remote
       - If a remote exists and Git has communicated with it (fetch/pull/push), 
         then Git creates remote-tracking branches for the branches it knows about on that remote.
```

<div style="text-align: center; color: blue;">
    <h1>Chapter 3: Getting changes down to local from github</h1>
</div>

<div style="text-align: center;">
    <h2>git fetch </h2 git restore
</div>

```

#### Forking 

```
Forking means creating your own copy of someone else’s repository on your GitHub account.

When you fork a repository:
    A copy of the project is created under your GitHub account
    You can freely experiment, modify, or improve the code
    The original repository remains unchanged
    You can later suggest your changes back using a Pull Request

original git repo     - github.com/original-user/project
after forking you get - github.com/your-username/project

As like PR, forking is not a Git feature.ability to fork is implemented on Github level
```

```
- Fork & clone workflow is different from anything we've seen so far
- Instead of just one centralized Github repository, every developer has their own Github repository 
  in addtion to the "main" repo.
- Developers make changes and push to their own forks before making PR
- It's very commonly used on large open-source projects where there may be thousands of contributors 
  with only a couple of maintainers
- The fork-and-clone workflow allows anyone with a GitHub account to contribute to an open-source project, 
  even if they are not directly listed as a collaborator by the maintainer or original repo.
  they can show their work just by forking and showing without being explicitly listed as contributor
- After forking you can invite your own favorite collaborator for contribution
- → After forking and cloning a repository, you should configure the original repository as an upstream remote.
  → In GitHub, the default remote origin points to your fork, not the original repository.
  → Your fork does not automatically receive updates from the original project.
  → To stay updated, add the original repository as upstream and periodically fetch and merge changes.
```
